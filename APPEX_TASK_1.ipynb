```python
import pandas as pd
df = pd.read_excel("ApexPlanet_DataAnalytics_Dataset.xlsx")
print("========== FIRST 5 ROWS ==========")
print(df.head())
```

```python
print("\n========== DATASET SHAPE ==========")
print(df.shape)

print("\n========== COLUMN NAMES ==========")
print(df.columns)

print("\n========== DATA TYPES ==========")
print(df.dtypes)

print("\n========== DATASET INFORMATION ==========")
print(df.info())

print("\n========== STATISTICAL SUMMARY ==========")
print(df.describe())
```

```python
print("\n========== MISSING VALUES ==========")
print(df.isnull().sum())
```

```python
duplicates = df.duplicated().sum()
print("\n========== DUPLICATE RECORDS ==========")
print("Duplicate Rows:", duplicates)
df = df.drop_duplicates()
numeric_columns = ["Age", "Quantity", "Unit_Price", "Total_Sales"]

for col in numeric_columns:
    if col in df.columns:
        df[col] = df[col].fillna(df[col].median())
categorical_columns = [
    "Gender",
        "City",
            "Product",
                "Category",
                    "Customer_Name"
                    ]
for col in categorical_columns:
      if col in df.columns:
              df[col] = df[col].fillna(df[col].mode()[0])
df["Order_Date"] = pd.to_datetime(df["Order_Date"], errors="coerce")
text_columns = [
    "Customer_Name",
        "Gender",
            "City",
                "Product",
                    "Category"
                    ]
```

```python
for col in text_columns:
      df[col] = df[col].astype(str).str.strip().str.title()
      df = df[(df["Age"] >= 18) & (df["Age"] <= 100)]
      df = df[df["Quantity"] > 0]
      df = df[df["Unit_Price"] > 0]
      df["Total_Sales"] = df["Quantity"] * df["Unit_Price"]
      print("\n========== CLEANED DATASET ==========")
      print(df.head())
      print("\n========== FINAL SHAPE ==========")
      print(df.shape)

      print("\n========== MISSING VALUES AFTER CLEANING ==========")
      print(df.isnull().sum())
      df.to_excel("ApexPlanet_DataAnalytics_Dataset_Cleaned.xlsx", index=False)

      print("\nDataset cleaned successfully!")
      print("Saved as: ApexPlanet_DataAnalytics_Dataset_Cleaned.xlsx")
```
