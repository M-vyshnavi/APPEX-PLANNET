```python
import pandas as pd
from scipy.stats import ttest_ind

# Load the Excel file
df = pd.read_excel("ApexPlanet_DataAnalytics_Dataset_Cleaned.xlsx")

# Display the first 5 rows
print("Dataset loaded successfully!")
display(df.head())
```

```python
male_sales = df[df["Gender"] == "Male"]["Total_Sales"].dropna()
female_sales = df[df["Gender"] == "Female"]["Total_Sales"].dropna()
print("Male Transactions:", len(male_sales))
print("Female Transactions:", len(female_sales))
```

```python
male_mean = male_sales.mean()
female_mean = female_sales.mean()

print("\nAverage Transaction Sales")
print("Male   :", round(male_mean, 2))
print("Female :", round(female_mean, 2))
```

```python
t_stat, p_value = ttest_ind(
    male_sales,
    female_sales,
    equal_var=False
)

print("\n========== HYPOTHESIS TEST RESULTS ==========")
print("T-statistic:", round(t_stat, 3))
print("P-value:", round(p_value, 3))
```

```python
alpha = 0.05

print("\nSignificance Level:", alpha)
```

```python
if p_value < alpha:
    print("\nDecision: Reject H0")
    print("Conclusion: There is a statistically significant")
    print("difference in average transaction sales between")
    print("Male and Female customers.")
else:
    print("\nDecision: Fail to Reject H0")
    print("Conclusion: There is NO statistically significant")
    print("difference in average transaction sales between")
    print("Male and Female customers.")
```
